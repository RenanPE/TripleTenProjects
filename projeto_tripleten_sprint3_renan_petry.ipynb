## Dicionário de dados

Há cinco tabelas no conjunto de dados, e você vai precisar usar todas elas para pré-processar seus dados e fazer AED. Abaixo está um dicionário que lista as colunas de cada tabela e descreve os dados contidos nelas.

- `instacart_orders.csv`: cada linha corresponde a um pedido no aplicativo da Instacart
    - `'order_id'`: é o número que identifica cada pedido de forma exclusiva
    - `'user_id'`: é o número de identificação exclusivo da conta de cada cliente
    - `'order_number'`: é o número de vezes que o cliente fez um pedido
    - `'order_dow'`: é o dia da semana em que o pedido foi feito (0 é domingo)
    - `'order_hour_of_day'`: é a hora do dia em que o pedido foi feito
    - `'days_since_prior_order'`: é o número de dias desde que o cliente fez seu pedido anterior




- `products.csv`: cada linha corresponde a um produto exclusivo que os clientes podem comprar
    - `'product_id'`: é o número de identificação unívoco de cada produto
    - `'product_name'`: é o nome do produto
    - `'aisle_id'`: é o número de identificação exclusivo de cada categoria de corredor do supermercado
    - `'department_id'`: é o número de identificação exclusivo de cada categoria de departamento do supermercado




-	`order_products.csv`: cada linha corresponde a um item incluído em um pedido
    -	`'order_id'`: é o número que identifica cada pedido de forma exclusiva
    -	`'product_id'`: é o número de identificação exclusivo de cada produto
    -	`'add_to_cart_order'`: é a ordem sequencial em que cada item foi colocado no carrinho
    -	`'reordered'`: 0 se o cliente nunca comprou o produto antes, 1 se já o comprou




-	`aisles.csv`
    -	`'aisle_id'`: é o número de identificação exclusivo de cada categoria de corredor do supermercado
    -	`'aisle'`: é o nome do corredor



-	`departments.csv`
    -	`'department_id'`: é o número de identificação exclusivo de cada categoria de departamento do supermercado
    -	`'department'`: é o nome do departamento

```python
# Há valores ausentes para os clientes que não estão fazendo o primeiro pedido?

## Solução: Verificar se há "user_id" com "days_since_prior_order" nulo, e que tenha outra ocorrência na base com "order_id" menor
df_orders_null = df_instacart_orders[df_instacart_orders['days_since_prior_order'].isna()]
df_aux_merge = df_instacart_orders[['order_id', 'user_id', 'days_since_prior_order']].rename(columns={'order_id': 'previous_order','days_since_prior_order':'days_since_prior_order_before'})

df_orders_null = df_orders_null.merge(df_aux_merge, how='inner', on='user_id').query("previous_order < order_id")

df_orders_null.sort_values(by=['user_id', 'order_id'])
```

```python
# df_instacart_orders[['order_id', 'user_id', 'days_since_prior_order']].merge(df_orders_null2, on=['order_id', 'user_id'], how='left', suffixes=('_a', '_b'))
```

```python
# Utilizando "merge_asof", para cruzar por valores proximados

qtd_days_since_prior_order_nulls = len(df_instacart_orders[df_instacart_orders['days_since_prior_order'].isna()])
print(f'Temos {qtd_days_since_prior_order_nulls} registros sem o campo de "days_since_prior_order" preenchido')

df_orders_null2 = df_instacart_orders[df_instacart_orders['days_since_prior_order'].isna()].sort_values(['order_id', 'user_id'])
df_orders_null2['aux_temp'] = 1  # Coluna auxiliar, para filtrar fora o que houver cruzamento

# filtrando valores identicos de "user_id" e "order_id", para cruzar por valores anteriores de "order_id" e iguais de "user_id"
df_aux_merge2 = df_instacart_orders[['order_id', 'user_id', 'days_since_prior_order']].merge(df_orders_null2[['order_id', 'user_id', 'aux_temp']], on=['order_id', 'user_id'], how='left', suffixes=('', '_b')).query("aux_temp.isna()").rename(columns={'order_id': 'previous_order','days_since_prior_order':'days_since_prior_order_before'}).sort_values(['previous_order', 'user_id'])
del df_orders_null2['aux_temp'] 
del df_aux_merge2['aux_temp']

df_merge_orders = pd.merge_asof(df_orders_null2, df_aux_merge2, left_on="order_id", right_on="previous_order", by="user_id", direction="backward").query("~ previous_order.isna()")
df_merge_orders['previous_order'] = df_merge_orders['previous_order'].astype('int64')
display(df_merge_orders)
# display(df_merge_orders.info())
```

### [C3] Para cada produto, qual parcela de todos os pedidos dele são repetidos?

```python
# Considerando como pedido repetido, o mesmo cliente comprando o mesmo produto mais de 1 vez:
cols_interesse = ['user_id', 'product_id', 'product_name']
df_grouped_cliente_product = df_merge.groupby(cols_interesse).size().reset_index(name='num_pedidos').sort_values(by='num_pedidos', ascending=False)
df_grouped_cliente_product.head(5)

df_repetidos = df_grouped_cliente_product[df_grouped_cliente_product['num_pedidos'] > 1]
df_repetidos['pedidos_repetidos'] = df_repetidos['num_pedidos'] - 1  # O 1° pedido do cliente e produto não é considerado repetido, apenas os próximos 

total_pedidos_por_produto = df_merge['product_id'].value_counts().reset_index(name='total_pedidos') #.to_frame(name='total_pedidos')
total_pedidos_por_produto.columns = ['product_id', 'total_pedidos']  

pedidos_repetidos_por_produto = df_repetidos.groupby(['product_id', 'product_name'])['pedidos_repetidos'].sum().reset_index()
pedidos_repetidos_por_produto = pedidos_repetidos_por_produto.sort_values(by='pedidos_repetidos', ascending=False)

# display(pedidos_repetidos_por_produto.head(6))
# display(total_pedidos_por_produto.head(5))

df_pedidos_repetidos_por_produto = pedidos_repetidos_por_produto.merge(total_pedidos_por_produto, how='left', on='product_id')
df_pedidos_repetidos_por_produto['proporcao'] = (df_pedidos_repetidos_por_produto['pedidos_repetidos'] / df_pedidos_repetidos_por_produto['total_pedidos']).round(3)


print('Top 10 produtos com maior proporção de repetidos')
display(df_pedidos_repetidos_por_produto.sort_values(by='proporcao', ascending=False).head(10))

print('\nTop 10 produtos com maior quantidade absoluta de pedidos repetidos')
display(df_pedidos_repetidos_por_produto.sort_values(by='pedidos_repetidos', ascending=False).head(10))

```

```python
# plotando a distribuição

# Criar o histograma
plt.figure(figsize=(14, 6))
hist = sns.histplot(df_pedidos_repetidos_por_produto['proporcao'], kde=True, bins=25)

plt.title('Distribuição da Proporção de Pedidos Repetidos por Produto', fontsize=14)
plt.xlabel('Proporção de Pedidos Repetidos', fontsize=12)
plt.ylabel('Frequência', fontsize=12)

# Exibir o gráfico
plt.show()
```
